The data present in the DataFrames can be accessed using indexing and slicing.

## INDEXING

**Indexing in DataFrames is of two types:**
 - Label based Indexing
 - Boolean Indexing.

![DataFrame Indexing Label](./images/dataframe-indexing.png)

```python
# Accessing using loc : row of index 2 or label
df.loc[2]
```

```python
# Accessing using iloc : the 2 index of the array 
df.iloc[2]
```

```python
df
```

```python
# Index values for both rows and columns
# Keep in mind slicing up to the len - 1
df.iloc[1:5 ,1:4] 
```

```python
df.loc[2:4 ,"Status":"PurchaseCost"]
```

```python
#The loc() method is used for label based indexing. It accepts row/column labels as parameters.
#Dataframe.loc[start_row:end_row,start_col:end_col]

#To display the row with row label 2, the command is:
df.iloc[2,:]

#where the symbol : indicates all columns.
```

```python
# To access columns: use:
# df.loc[:,start column:end column] Make sure not to miss the COLON BEFORE COMMA.
df.loc[:,"Status"]
```

**We can easily retrieve a slice of rows and columns using loc() and iloc() methods.**

The loc[] method is used to retrieve the group of rows and columns by labels or a boolean array present in the DataFrame. It takes only index labels, and if it exists in the called DataFrame, it returns the rows, columns, or DataFrame.

**The .iloc[] method is used retrieve the group of rows and columns at particular positions in the index so it only takes integers. It is used when the index label of the DataFrame **is other than numeric series of 0,1,2,	,n, or in the case when the user
does not know the index label

![DataFrame iloc](./images/df_iloc.webp)

```python
# Every 2nd record and every alternate column
df.iloc[::2,::2]
```

```python
# Rows of index 0 up to 2 and Columns from index 1 up to 2
df.iloc[0:3,1:3]
```

# Fastest Finger First

```python
# Books with FinalCost Higher than PurchaseCost
df[df["FinalCost"]>df["PurchaseCost"]]
```

```python
# Fetch all the books where Category is not "Cookery" (use ~ or !=).
df[df["Category"]!="Cookery"]

# or

# df[~(df["Category"] == "Cookery")]
```

```python
# Fetch rows for the 3 books with the highest PurchaseCost (use .nlargest()).
df.nlargest(3, "PurchaseCost")
```

```python
# Fetch books where PurchaseCost is between 15–25 and Status is "In".
df[(df["PurchaseCost"].between(15, 25)) & (df["Status"] == "In")]
```

```python
# Compute average PurchaseCost and fetch books costing more than average.
avg_cost = df["PurchaseCost"].mean()
df[df["PurchaseCost"] > avg_cost]
```
