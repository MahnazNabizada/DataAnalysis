# From Data to Business Decision 

```python
# Importing libraries
import pandas as pd
import numpy as np
```

```python
# Loading excel file as a dataframe
df = pd.read_excel("Book.xlsx")
#print(df.columns)
#print(df[["Customer_ID "]])
df
```

```python
# Remove completely empty rows
# Replacing extra whitespaces '\xa0' for empty string Ex: 'Customer_ID\xa0' -> 'Customer_ID '
df = df.map(
    lambda x: x.replace("\xa0", " ").strip()
    if isinstance(x, str)
    else x
)

# Convert empty strings to NaN
df = df.replace(r"^\s*$", np.nan, regex=True)

# Remove rows where every column is missing
# The reset_index(drop=True) is useful because after removing row 12, pandas may otherwise preserve the original index.
df = df.dropna(how='all').reset_index(drop=True)

df
```

```python
# Converting columns types
# Text / categorical columns
df["Customer_ID"] = df["Customer_ID"].astype("string")
df["Gender"] = df["Gender"].astype("string")
df["City"] = df["City"].astype("string")

# Numeric columns
df["Age"] = pd.to_numeric(df["Age"], errors="coerce").astype("Int64")
df["Income"] = pd.to_numeric(df["Income"], errors="coerce")
df["Visits"] = pd.to_numeric(df["Visits"], errors="coerce").astype("Int64")
df["Avg_Spend"] = pd.to_numeric(df["Avg_Spend"], errors="coerce")

# Date
df["Signup_Date"] = pd.to_datetime(df["Signup_Date"], errors="coerce")

df.dtypes
```

## Part 2 — Data Wrangling

### Q1 - Which columns contain missing values, and how many are missing in each? 

```python
# Coung missing values per column
print(df.isnull().sum())

# Check if any missing values exist
# df.isnull().any()

# Total missing values
print("Total number of missing values" ,df.isnull().sum().sum())  
```

### Q2 - How would you handle the missing Income value? Explain your choice. 

```python
# Replacing missing income value for the average
# I would replace the missing Income with the median income rather than dropping the row. 
# Since only one value is missing and the dataset is small, imputation preserves the customer record. 
# The median is preferable because income can contain outliers that could distort the mean.
df["Income"] = df["Income"].fillna(df["Income"].median())
df
```

### Q3 - How would you standardize the Gender values so that F, female, and Female are consistent? 

```python
standardize = {"female": "F", "f": "F", "male": "M", "m": "M"}

df["Gender"] = df["Gender"].str.strip().str.lower()
df["Gender"] = df["Gender"].replace(standardize)

df
```

### Q4 - Are there duplicate customer records? Identify and handle them appropriately. 

```python
# Checking for entire duplicated rows
print(f"Duplicated rows : {df.duplicated().sum()}")

# For our data, C002 and C011 would not be considered duplicates because their Customer_ID values differ.
# Check duplicates excluding Customer_ID
print(f"Duplicated excluding Customer_ID : {df.duplicated(subset=df.columns.difference(["Customer_ID"])).sum()}")

# Investigating before deleting
print(df[df["Customer_ID"].isin(["C002", "C011"])])

# Droping the duplicated records keeping first occurrence
df = df.drop_duplicates(
    subset=df.columns.difference(["Customer_ID"]),
    keep="first"
).reset_index(drop=True)
df
```
