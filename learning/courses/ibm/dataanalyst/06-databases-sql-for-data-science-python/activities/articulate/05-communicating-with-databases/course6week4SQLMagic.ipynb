#Date & time function for SQlite

| Task          | SQLite                  | SQL Server             |
| ------------- | ----------------------- | ---------------------- |
| Current date  | `DATE('now')`           | `GETDATE()`            |
| Extract year  | `strftime('%Y',date)`   | `YEAR(date)`           |
| Extract month | `strftime('%m',date)`   | `MONTH(date)`          |
| Add days      | `DATE(date,'+10 days')` | `DATEADD(day,10,date)` |
| Difference    | `julianday()`           | `DATEDIFF()`           |


```python
%%sql

CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY,
    customer TEXT,
    order_date DATE,
    amount REAL
);
```

```python
%%sql

INSERT INTO orders (order_id, customer, order_date, amount)
VALUES
(1, 'John', '2026-01-15', 100.00),
(2, 'Sarah', '2026-02-20', 250.00),
(3, 'Mike', '2026-08-01', 150.00),
(4, 'Emma', '2025-12-10', 300.00),
(5, 'David', '2026-07-15', 200.00);
```

```python
%%sql

SELECT *
FROM orders;
```

```python
%%sql

SELECT
    customer,
    order_date
FROM orders
where  strftime('%Y', order_date)='2026';
```

```python
%%sql

SELECT *
FROM orders
WHERE strftime('%Y', order_date) = '2026';
```

**Last 30 days**

```python
%%sql

SELECT *
FROM orders
WHERE order_date >= DATE('now','-30 days');
```

**Monthly sales**

```python
%%sql

SELECT
    strftime('%Y-%m', order_date) AS month,
    SUM(amount) AS total_sales
FROM orders
GROUP BY month;
```
