### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset into a DataFrame



load the dataset using pd.read_csv()


```python
# Define the URL of the dataset
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3: Identifying Duplicate Rows


**Task 1: Identify Duplicate Rows**
  1. Count the number of duplicate rows in the dataset.
  2. Display the first few duplicate rows to understand their structure.


```python
## Write your code here
duplicate_count = df.duplicated().sum()
duplicates = df[df.duplicated(keep=False)]

print(f"Number of duplicated rows: {duplicate_count}")
duplicates.head()
```

### Step 4: Removing Duplicate Rows


**Task 2: Remove Duplicates**
   1. Remove duplicate rows from the dataset using the drop_duplicates() function.
2. Verify the removal by counting the number of duplicate rows after removal .


```python
# Remove duplicates by default keep the first occurrence and removes subsequent identical rows
df_clean = df.drop_duplicates()

# Verify the removal
duplicate_count_after = df_clean.duplicated().sum()
print(f"Number of duplicated rows after removal: {duplicate_count_after}")

# Compare the row counts
rows_before = len(df)
rows_after = len(df_clean)
rows_removed = rows_before - rows_after

# Reporting rows counts
print(f"Rows before: {rows_before}")
print(f"Rows after: {rows_after}")
print(f"Rows removed: {rows_removed}")
print(f"Duplicates after removal: {df_clean.duplicated().sum()}")
```

### Step 5: Handling Missing Values


**Task 3: Identify and Handle Missing Values**
   1. Identify missing values for all columns in the dataset.
   2. Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value.


```python
# Identifying missing values for all columns
missing_values = df_clean.isnull().sum()

# Calculating the percentage of missing values
missing_summary = pd.DataFrame({
    "Missing Values": df_clean.isnull().sum(),
    "Missing Percentage": df_clean.isnull().mean() * 100
})

missing_summary = missing_summary.sort_values(
    "Missing Percentage",
    ascending=False
)

missing_summary.head(10)
```

```python
# Check missing values in EdLevel
edlevel_missing_values = df_clean["EdLevel"].isnull().sum()
print(f"Number of missing values in EdLevel: {edlevel_missing_values}")

# Checking the EdLevel existing values
df_clean["EdLevel"].value_counts(dropna=False)
```

```python
# Find the most frequent EdLevel to inpute in missing values
edlevel_mode = df_clean["EdLevel"].mode()[0]
print(f"Most frequent EdLevel:", edlevel_mode)

# Input missing EdLevel values
df_clean["EdLevel"] = df_clean["EdLevel"].fillna(edlevel_mode)

# Verify missing values after inputing
edlevel_missing_values_after = df_clean["EdLevel"].isnull().sum()
print(f"Number of missing values in EdLEvel after inputing: {edlevel_missing_values_after}")
```

### Step 6: Normalizing Compensation Data


**Task 4: Normalize Compensation Data Using ConvertedCompYearly**
   1. Use the ConvertedCompYearly column for compensation analysis as the normalized annual compensation is already provided.
   2. Check for missing values in ConvertedCompYearly and handle them if necessary.


```python
# Checking the ConvertedCompYearly stats
df_clean["ConvertedCompYearly"].describe()
```

```python
# Checking for missing values in ConvertedCompYearly
missing_comp = df_clean["ConvertedCompYearly"].isnull().sum()
print(f"Missing ConvertedCompYearly values: {missing_comp}")

missing_comp_pct = df_clean["ConvertedCompYearly"].isnull().mean() * 100
print(f"Missing percentage: {missing_comp_pct:.2f}%")
```

```python
# Create a dataset for compensation analysis
compensation_data = df_clean.dropna(subset=["ConvertedCompYearly"])

# Cheking missing values after handling
missing_comp_after = compensation_data["ConvertedCompYearly"].isnull().sum()
print(f"Missing values after handling: {missing_comp_after}")
```
