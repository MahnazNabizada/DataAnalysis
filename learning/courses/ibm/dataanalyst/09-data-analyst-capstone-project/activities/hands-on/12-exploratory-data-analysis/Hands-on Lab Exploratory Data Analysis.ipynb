#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
# Identifying critical columns
critical_columns = ["Employment", "JobSat", "RemoteWork"]

# Identifying missing values for all columns
missing_values = df[critical_columns].isnull()

missing_summary = pd.DataFrame({
    "Missing Values": missing_values.sum(),
    "Missing Percentage": (
        missing_values.mean() * 100
    ).round(2)
})

display(missing_summary)
```

```python
# Inspecting the distributions
for column in critical_columns:
    print(f"\n{column}")
    print(df[column].value_counts(dropna=False))
```

```python
# For RemoteWork - 16.25% missing values
# There are a moderate amount of missing data. Since RemoteWork is categorical I'll use mode imputation
remote_work_mode = df["RemoteWork"].mode()[0]
df["RemoteWork"] = df["RemoteWork"].fillna(remote_work_mode)

# Verifying missing values after imputation
print(f"Missing values after imputation: {df["RemoteWork"].isnull().sum()}")
```

```python
# For JobStat - 55.49% missing values
# This is not recommended mode imputation. More than half of the observations are missing
# If we did df["JobSat"] = df["JobSat"].fillna(df["JobSat"].mode()[0]) you would be assigning the same satisfaction level to 36,311 respondents who didn't provide a response
# That could substantially distort your analysis
# It'S recommended keeping the missing values as NaN and excluding them only when you're analyzing job satisfaction
```

**Conclusion:** Employment contained no missing values, so no action was required. RemoteWork had 16.25% missing values and was imputed using the mode because it is a categorical variable. JobSat had 55.49% missing values, so mode imputation was avoided to prevent significant distortion of the distribution. Missing JobSat values were retained and will be excluded only from analyses requiring job satisfaction responses.

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
# The recommended approch for this is using pd.cut() to create experience groups and then groupby().median() to calculate the median JobStat
# Making YearsCodePro numeric
df["YearsCodePro"] = df["YearsCodePro"].replace({
    "Less than 1 year": "0",
    "More than 50 years": "51"
})

df["YearsCodePro"] = pd.to_numeric(
    df["YearsCodePro"],
    errors="coerce"
)

# Creating ranges for YearsCodePro
bins = [-1, 5, 10, 20, float("inf")]
labels = ["0-5", "5-10", "10-20", ">20"]

df["ExperienceRange"] = pd.cut(
    df["YearsCodePro"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
# Calculating median JobSat by experience
# JobSat is already numeric we can calculate the median
median_jobsat = (
    df.groupby("ExperienceRange", observed=True)["JobSat"]
      .median()
)

median_jobsat
```

```python
# Including the number of valid responses, which I strongly recommend because 55.49% of JobSat is missing
jobsat_summary = (
    df.groupby("ExperienceRange", observed=True)
      .agg(
          MedianJobSat=("JobSat", "median"),
          ValidResponses=("JobSat", "count")
      )
)

display(jobsat_summary)
```

```python
# Creating the Bar plot
plt.figure(figsize=(8, 5))

sns.barplot(
    data=jobsat_summary.reset_index(),
    x="ExperienceRange",
    y="MedianJobSat"
)

plt.title("Median Job Satisfaction by Experience Range")
plt.xlabel("Professional Coding Experience (Years)")
plt.ylabel("Median Job Satisfaction (0-10)")

plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
# For this task, the main challenge is that LanguageHaveWorkedWith contains multiple languages in a single cell
# Example: Python;JavaScript;SQL;HTML/CSS
# First we need to split the languages and count each language separately

# Inspecting LanguageHaveWorkedWith
df["LanguageHaveWorkedWith"].head()
```

```python
# Selecting the top 10 countries by respondent count
top_countries = df["Country"].value_counts().head(10).index

df_region = df[
    df["Country"].isin(top_countries)
].copy()

top_countries
```

```python
# Spliting the programming langugaes
languages = (
    df_region["LanguageHaveWorkedWith"]
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
)

languages
```

```python
# Match each split language with its respondent's country
language_data = pd.DataFrame({
    "Country": df_region.loc[languages.index, "Country"].values,
    "Language": languages.values
})

language_data.head(20)
```

```python
# Cross tabulating Top 10 countries by Programming Languages
language_country = pd.crosstab(
    language_data["Country"],
    language_data["Language"]
)

language_country
```

```python
# Identifying the top 10 languages
top_languages = (
    language_data["Language"]
    .value_counts()
    .head(10)
)

top_languages
```

```python
# Cross tabulating Top 10 countries by the top 10 Programming Languages
language_country_top10 = language_country[
    top_languages.index
]

language_country_top10
```

```python
# Since your 10 countries don't necessarily have the same number of respondents, percentages are more meaningful than raw counts
language_country_pct = (
    language_country_top10
    .div(language_country.sum(axis=1), axis=0)
    * 100
)

language_country_pct = language_country_pct.round(1)

language_country_pct
```

```python
# Generating a heatmap for a better data visualization
plt.figure(figsize=(12, 7))

sns.heatmap(
    language_country_pct,
    annot=True,
    fmt=".1f"
)

plt.title("Top 10 Programming Languages by Country")
plt.xlabel("Programming Language")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```
