### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
# Identifying missing values on critical columns
critical_columns = ["Employment", "JobSat", "RemoteWork"]

missing_values = (
    df[critical_columns].isnull()
      .sum()
      .sort_values(ascending=False)
)
missing_values
```

```python
# Calculating the percentage of missing values
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean() * 100).round(2)
})

missing_summary = missing_summary.sort_values(
    "Missing Percentage",
    ascending=False
)

missing_summary
```

**Recommended** approach for handling missing values per situation:

| Situation                                | Recommended approach                                      |
|------------------------------------------|-----------------------------------------------------------|
| Very few missing values                  | Impute or remove rows                                     |
| Numerical column                         | Median or mean                                            |
| Categorical column                       | Mode or "Unknown"                                         |
| Critical variable with many missing values | Consider keeping missing values or dropping the column  |
| Extremely high missing percentage        | Consider excluding the column from analysis               |


```python
# For example, if ConvertedCompYearly is missing, you could use the median because compensation tends to be highly skewed:
df["ConvertedCompYearly"] = df["ConvertedCompYearly"].fillna(
    df["ConvertedCompYearly"].median()
)
```

```python
# For a categorical variable such as RemoteWork, you could use the most frequent value
df["RemoteWork"] = df["RemoteWork"].fillna(
    df["RemoteWork"].mode()[0]
)
```

**Important:** For JobSat I wouldn't automatically fill JobSat with its median. If more than half of the respondents didn't provide a satisfaction score, replacing all those missing responses with the median would artificially change the distribution. Instead, you can leave them as NaN and exclude them when analyzing JobSat

```python
# Checking missing values afterwards
missing_after = (
    df[critical_columns].isnull()
      .sum()
      .sort_values(ascending=False)
)

display(missing_after)
```

**Conclusion:** Missing values were identified and evaluated based on their frequency and data type. Numerical values were imputed using appropriate statistics such as the median, while categorical values were handled using the most frequent value when appropriate. Columns with substantial missing data were not blindly imputed to avoid introducing bias into the analysis. The dataset was then checked again to verify the missing-value treatment.

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
# Since JobSat is already numeric on a 0–10 scale, a KDE plot is a good choice for showing the overall distribution
# Create the KDE plot
# As we are working on JobSat now we can remove the missing values
jobsat_data = df["JobSat"].dropna()

# Generating a KDE plot
plt.figure(figsize=(10, 6))

sns.kdeplot(
    data=jobsat_data,
    fill=True
)

plt.title("Distribution of Job Satisfaction")
plt.xlabel("Job Satisfaction Score")
plt.ylabel("Density")

plt.xlim(0, 10)

plt.show()
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
# For this task, the key issue is that both columns contain multiple programming languages in each row, separated by ;
# I recommend a grouped bar chart rather than a Venn diagram. A Venn diagram becomes difficult to interpret with many programming languages

# Split and count the languages
worked_languages = (
    df["LanguageHaveWorkedWith"]
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
)

wanted_languages = (
    df["LanguageWantToWorkWith"]
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
)

worked_counts = worked_languages.value_counts()
wanted_counts = wanted_languages.value_counts()

worked_counts
wanted_counts
```

```python
# Creating a data frame combining both counts
language_comparison = pd.DataFrame({
    "Worked": worked_counts,
    "Want": wanted_counts
}).fillna(0)

language_comparison.head()
```

```python
# Selecting top 10 languages based on combined frequency
language_comparison["Total"] = (
    language_comparison["Worked"] +
    language_comparison["Want"]
)

top_languages = (
    language_comparison
    .sort_values("Total", ascending=False)
    .head(10)
)

# Removing temporary Total column
top_languages = top_languages.drop(columns="Total")

top_languages
```

```python
# Creating a grouped bar chart
top_languages.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Programming Languages: Worked With vs. Want to Work With")
plt.xlabel("Programming Language")
plt.ylabel("Number of Respondents")

plt.xticks(rotation=45, ha="right")

plt.legend(
    title="Language Preference"
)

plt.tight_layout()
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
# For this task, I recommend using Country as the region and a heatmap of percentages
# Since the dataset has many countries, showing every country would make the visualization difficult to read.
# Proper workflow is: Country → RemoteWork → select major countries → calculate percentages → heatmap

# Check the RemoteWork distribution
df["RemoteWork"].value_counts(dropna=False)
```

```python
# Selecting the top 10 respondent's countries
top_countries = (
    df["Country"]
    .value_counts()
    .head(10)
    .index
)

# Copy all records from the top 10 respondent's countries to df_region dataframe
df_region = df[
    df["Country"].isin(top_countries)
].copy()

top_countries
```

```python
# Creating a cross-tabulation 
# First checing raw counts
remote_country = pd.crosstab(
    df_region["Country"],
    df_region["RemoteWork"]
)

remote_country
```

```python
# Converting counts to percentages
remote_country_pct = (
    pd.crosstab(
        df_region["Country"],
        df_region["RemoteWork"],
        normalize="index"
    ) * 100
).round(2)

remote_country_pct
```

```python
# Creating a heatmap chart for better visualization
plt.figure(figsize=(10, 7))

sns.heatmap(
    remote_country_pct,
    annot=True,
    fmt=".1f"
)

plt.title("Remote Work Distribution by Country")
plt.xlabel("Remote Work Arrangement")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
# You can also create a grouped bar chart 
remote_country_pct.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Remote Work Distribution by Country")
plt.xlabel("Country")
plt.ylabel("Percentage of Respondents")

plt.xticks(rotation=45, ha="right")
plt.legend(title="Remote Work")

plt.tight_layout()
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
# For this task, the main issue is that YearsCodePro may not be numeric
# It can contain values such as "Less than 1 year" and "More than 50 years". So, before calculating a correlation, we need to convert it appropriately
# Selecting relevant columns
correlation_data = df[["YearsCodePro", "JobSat"]].copy()
correlation_data.info()
```

```python
# Converting YearsCodePro to numeric
correlation_data["YearsCodePro"] = pd.to_numeric(
    correlation_data["YearsCodePro"],
    errors="coerce"
)

correlation_data["YearsCodePro"].info()
```

```python
# Removing missing values
correlation_data = correlation_data.dropna(
    subset=["YearsCodePro", "JobSat"]
)
correlation_data.head()
```

```python
# Calculating Pearson correlation
pearson_corr = correlation_data["YearsCodePro"].corr(
    correlation_data["JobSat"],
    method="pearson"
)

print(f"Pearson correlation: {pearson_corr:.3f}")
```

```python
# Calculating Spearman correlation
spearman_corr = correlation_data["YearsCodePro"].corr(
    correlation_data["JobSat"],
    method="spearman"
)

print(f"Spearman correlation: {spearman_corr:.3f}")
```

```python
# Visualizaing the relationship using scatter plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=correlation_data,
    x="YearsCodePro",
    y="JobSat",
    alpha=0.3
)

plt.title("Job Satisfaction vs. Professional Coding Experience")
plt.xlabel("Years of Professional Coding Experience")
plt.ylabel("Job Satisfaction")

plt.tight_layout()
plt.show()
```

**Conclusion:** The Pearson correlation between professional coding experience and job satisfaction is 0.104, while the Spearman correlation is 0.119. Both results indicate a very weak positive relationship, suggesting that respondents with more professional coding experience tend to report slightly higher job satisfaction. However, the strength of the relationship is low, so experience alone is not a strong predictor of job satisfaction

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
# For this task, the best approach is to use pd.crosstab() for counts, then convert it to row percentages for a more meaningful comparison. Since Employment has many categories in this dataset, a stacked bar chart with every category would be very crowded, so I'd limit the visualization to the most common employment types.
# Preparing the Data - removing rows where either Employment or EdLevel is missing
employment_education = df[
    ["Employment", "EdLevel"]
].dropna()

print("Rows used for analysis:", len(employment_education))
```

```python
# Create the cross-tabulation
employment_education_counts = pd.crosstab(
    employment_education["Employment"],
    employment_education["EdLevel"]
)

employment_education_counts
```

```python
# Calculate percentages
employment_education_pct = pd.crosstab(
    employment_education["Employment"],
    employment_education["EdLevel"],
    normalize="index"
) * 100

employment_education_pct = employment_education_pct.round(2)

employment_education_pct
```

```python
# Selecting the most common employment types
top_employment = (
    employment_education["Employment"]
    .value_counts()
    .head(10)
    .index
)

top_employment_pct = employment_education_pct.loc[
    top_employment
]

top_employment_pct
```

```python
# Create the stacked bar plot
top_employment_pct.plot(
    kind="bar",
    stacked=True,
    figsize=(14, 8)
)

plt.title("Education Level Distribution by Employment Type")
plt.xlabel("Employment Type")
plt.ylabel("Percentage of Respondents")

plt.xticks(rotation=45, ha="right")

plt.legend(
    title="Education Level",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```

```python
# let's look into a different perspective:
# What employment types are represented within each education level?
education_employment_pct = pd.crosstab(
    employment_education["EdLevel"],
    employment_education["Employment"],
    normalize="index"
) * 100

top_employment_education = education_employment_pct[
    top_employment
]
```

```python
# Visualizing the transposed analysis
top_employment_education.plot(
    kind="bar",
    stacked=True,
    figsize=(14, 8)
)

plt.title("Employment Type Distribution by Education Level")
plt.xlabel("Education Level")
plt.ylabel("Percentage of Respondents")

plt.xticks(rotation=45, ha="right")

plt.legend(
    title="Employment Type",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```

**Conclusion:** The cross-tabulation shows differences in educational backgrounds across employment types. The stacked bar chart indicates that some employment categories have a higher proportion of respondents with bachelor's or master's degrees, while other categories contain larger proportions of respondents with lower or different levels of formal education. These patterns suggest an association between education level and employment type, although the cross-tabulation does not establish a causal relationship
