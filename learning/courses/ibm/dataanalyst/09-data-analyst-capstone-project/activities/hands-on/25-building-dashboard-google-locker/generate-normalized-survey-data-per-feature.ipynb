```python
# Importing required libraries
import pandas as pd
```

```python
# Read the csv file into a pandas dataframe
df = pd.read_csv('survey_data_updated.csv')

# Display the first few rows of the data
df.head()
```

```python
# Check the columns
df.describe()

# Cleaning spaces from column names
df.columns = (
    df.columns
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)
```

```python
summary = pd.DataFrame({
    'Data Type': df.dtypes,
    'Unique Values': df.nunique(dropna=False),
    'Missing Values': df.isna().sum()
})

summary
```

```python
pd.set_option('display.max_rows', None)

summary
```

```python
# Let's split the LanguageWantToWorkWith, LanguagesHaveWorkedWith per value counts
language_want_work_counts = (
    df["LanguageWantToWorkWith"]
    .dropna()
    .str.split(";")
    .explode()
    .value_counts()
)

language_have_worked_counts = (
    df["LanguageHaveWorkedWith"]
    .dropna()
    .str.split(";")
    .explode()
    .value_counts()
)

databases_have_worked_counts = (
    df["DatabaseHaveWorkedWith"]
    .dropna()
    .str.split(";")
    .explode()
    .value_counts()
)

platforms_have_worked_counts = (
    df["PlatformHaveWorkedWith"]
        .dropna()
        .str.split(";")
        .explode()
        .value_counts()
)

webframeworks_have_worked_counts = (
    df["WebframeHaveWorkedWith"]
        .dropna()
        .str.split(";")
        .explode()
        .value_counts()
)

# Split WebframeHaveWorkedWith and create one row per platform
webframeworks_have_worked_df = (
    df[["ResponseId", "WebframeHaveWorkedWith"]]
    .dropna(subset=["WebframeHaveWorkedWith"])
    .assign(WebframeHaveWorkedWith=lambda x: x["WebframeHaveWorkedWith"].str.split(";"))
    .explode("WebframeHaveWorkedWith")
)

# Reset index
webframeworks_have_worked_df = webframeworks_have_worked_df.reset_index(drop=True)
webframeworks_have_worked_df.insert(0, "Id", range(1, len(webframeworks_have_worked_df) + 1))

# Find the top 10 web frameworks
top_10_web_frameworks = (
    webframeworks_have_worked_df["WebframeHaveWorkedWith"]
    .value_counts()
    .head(10)
)

# Keep only records belonging to the top 10 platforms
top_10_web_frameworks_df = webframeworks_have_worked_df[
    webframeworks_have_worked_df["WebframeHaveWorkedWith"].isin(top_10_web_frameworks.index)
]

languages_want_work_with_counts = (
    df["LanguageWantToWorkWith"]
        .dropna()
        .str.split(";")
        .explode()
        .value_counts()
)

databases_want_work_wth_counts = (
    df["DatabaseWantToWorkWith"]
    .dropna()
    .str.split(";")
    .explode()
    .value_counts()
)

platforms_want_work_with_counts = (
    df["PlatformWantToWorkWith"]
    .dropna()
    .str.split(";")
    .explode()
    .value_counts()
)

# Split WebframeWantToWorkWith and create one row per platform
webframeworks_want_work_with_df = (
    df[["ResponseId", "WebframeWantToWorkWith"]]
    .dropna(subset=["WebframeWantToWorkWith"])
    .assign(WebframeWantToWorkWith=lambda x: x["WebframeWantToWorkWith"].str.split(";"))
    .explode("WebframeWantToWorkWith")
)

# Reset index
webframeworks_want_work_with_df = webframeworks_want_work_with_df.reset_index(drop=True)
webframeworks_want_work_with_df.insert(0, "Id", range(1, len(webframeworks_want_work_with_df) + 1))

# Find the top 10 web frameworks want to work
top_10_web_frameworks_want_work_with = (
    webframeworks_want_work_with_df["WebframeWantToWorkWith"]
    .value_counts()
    .head(10)
)

# Keep only records belonging to the top 10 web frameworks respondents want to work with
top_10_web_frameworks_want_work_with_df = webframeworks_want_work_with_df[
    webframeworks_want_work_with_df["WebframeWantToWorkWith"].isin(top_10_web_frameworks_want_work_with.index)
]

top_10_web_frameworks_want_work_with_df.head(20)
```

```python
# And save in a new file for our analysis
language_have_worked_counts.to_csv("survey-data-languages-have-worked-with.csv")
databases_have_worked_counts.to_csv("survey-data-databases-have-worked-with.csv")
platforms_have_worked_counts.to_csv("survey-data-platforms-have-worked-with.csv")
top_10_web_frameworks_df.to_csv("survey-data-webframeworks-have-worked-with.csv", index=False)
languages_want_work_with_counts.to_csv("survey-data-languages-want-work-with.csv")
databases_want_work_wth_counts.to_csv("survey-data-databases-want-work-with.csv")
platforms_want_work_with_counts.to_csv("survey-data-platforms-want-work-with.csv")
top_10_web_frameworks_want_work_with_df.to_csv("survey-data-webframeworks-want-work-with.csv", index=False)
```
