```python
# Importing required libraries
import pandas as pd
```

```python
# Read the csv file into a pandas dataframe
df = pd.read_csv('survey_data_updated.csv')

# Display the first few rows of the data
df.head()
```

```python
# Let's split the LanguageWantToWorkWith, LanguagesHaveWorkedWith per value counts
language_want_work_counts = (
    df["LanguageWantToWorkWith"]
    .dropna()
    .str.split(";")
    .explode()
    .value_counts()
)

language_have_worked_counts = (
    df["LanguageHaveWorkedWith"]
    .dropna()
    .str.split(";")
    .explode()
    .value_counts()
)

language_want_work_counts
language_have_worked_counts
```

```python
# And save in a new file for our analysis
language_want_work_counts.to_csv("survey-data-languages-want-work-with.csv")
language_have_worked_counts.to_csv("survey-data-languages-have-worked-with.csv")
```
