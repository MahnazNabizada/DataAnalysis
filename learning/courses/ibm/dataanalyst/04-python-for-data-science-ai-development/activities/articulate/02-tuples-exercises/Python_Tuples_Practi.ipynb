```python
#Q19 check whether Coursera exists within  myTuple
def contains_element(nested, target):
  for item in nested:
    if isinstance(item, tuple):
      if contains_element(item, target):
        return True
    elif item == target:
      return True
  return False

found = contains_element(nestedTuple, "Tuple")
found

# NOTE "in" doesn't work properly with nested tuples
```

```python
#Q21 Find the index of JDA in myTuple
myTuple.index("JDA")

# Find the index of 'Coursera' in nestedTuple
def find_index(nested, target, path=()):
  for index, item in enumerate(nested):
    current_path = path + (index,)

    if item == target:
      return current_path

    if isinstance(item, tuple):
      result = find_index(item, target, current_path)
      if result:
        return result
  return None

print(find_index(nestedTuple, "week"))

# NOTE index doesn't work properly with nested tuples # Wrong output
```
