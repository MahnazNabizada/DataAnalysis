#List Data Structure

```python
# 1. Creating Lists
numbers = [10, 20, 30, 40, 50]               # List of integers
fruits = ["apple", "banana", "mango"]        # List of strings
mixed = [1, "hello", 3.14, True]             # Mixed data types
nested = [[1, 2], [3, 4], [5, 6]]            # Nested list (list of lists)

print("Numbers:", numbers)
print("Fruits:", fruits)
print("Mixed:", mixed)
print("Nested:", nested)
print("-------------------------------------------------")

# 2. Accessing Elements
print("First number:", numbers[0])           # Indexing (0-based)
print("Last fruit:", fruits[-1])             # Negative indexing
print("Second element in nested list:", nested[1][0])  # Accessing nested elements
print("-------------------------------------------------")

# 3. Slicing
L=[23,34,45,67,66,89]
print("First 3 numbers:", numbers[0:3])      # Slicing [start:end]
print("From index 2 to end:", numbers[2:])   # Omitting end means till last
print("Every 2nd number:", numbers[::2])     # Step slicing
print("-------------------------------------------------")

# 4. Mutability (Lists can be modified)
print("Before update:", fruits)
fruits[1] = "orange"                         # Updating element
print("After update:", fruits)
print("-------------------------------------------------")

# 5. Adding Elements
fruits.append("grape")                       # Append single item
print("After append:", fruits)

fruits.extend(["kiwi", "melon"])            # Extend (add multiple items)
print("After extend:", fruits)

fruits.insert(1, "strawberry")               # Insert at specific position
print("After insert:", fruits)
print("-------------------------------------------------")

# 6. Removing Elements
fruits.remove("orange")                      # Remove by value
print("After remove:", fruits)

popped_item = fruits.pop()                   # Pop last item
print("Popped item:", popped_item)
print("After pop:", fruits)

del fruits[0]                                # Delete by index
print("After del:", fruits)
print("-------------------------------------------------")

# 7. Checking length
print("Number of fruits:", len(fruits))

# 8. Looping through list
print("Looping through numbers:")
for num in numbers:
    print(num)
print("-------------------------------------------------")

# 9. Clearing the list
numbers.clear()
print("Numbers after clear:", numbers)
```

#What is a Dictionary?
A dictionary is a collection of key-value pairs.

Keys must be unique and immutable (e.g., strings, numbers, tuples).

Values can be of any type.

Defined using curly braces {}

```python
student = {
    "name": "Alice",
    "age": 20,
    "course": "Data Analytics"
}


print(student["name"])      # Alice
print(student.get("age"))   # 20


student["email"] = "alice@example.com"  # Add new key-value pair
student["age"] = 21                     # Update existing value
print(student)


student.pop("course")  # Remove a key by name
del student["email"]   # Delete a key
print(student)
```

```python
print(student.keys())    # dict_keys(['name', 'age'])
print(student.values())  # dict_values(['Alice', 21])
print(student.items())   # dict_items([('name', 'Alice'), ('age', 21)])
```

#Numpy

What is NumPy?
NumPy (Numerical Python) is a Python library used for fast numerical computations.

It provides the ndarray (N-dimensional array) data structure.

It’s more efficient than lists for mathematical operations and is widely used in data analysis and machine learning.

Key Features
Fast performance: Vectorized operations (no loops needed).

Supports multi-dimensional arrays: 1D, 2D (matrix), 3D, etc.

Built-in mathematical functions: sum, mean, sqrt, etc.

Easy integration: Works with Pandas, Matplotlib, and other libraries.

```python
import numpy as np

# 1D array
arr1 = np.array([1, 2, 3, 4, 5])
print("1D Array:", arr1)

# 2D array (matrix)
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
print("\n2D Array:\n", arr2)
```

A NumPy array is homogeneous—it stores elements of the same data type. While it is primarily used for numerical computations, it can also store strings, booleans, dates, or objects.
NumPy arrays are stored in contiguous memory.
Most operations are implemented in optimized C code, so they avoid the overhead of Python loops.
This makes NumPy significantly faster, especially for large datasets.

```python
import numpy as np

# Integers
a = np.array([1, 2, 3])

# Floats
b = np.array([1.5, 2.8, 3.2])

# Strings
c = np.array(["apple", "banana", "orange"])

# Booleans
d = np.array([True, False, True])
```

#What is Pandas?
Pandas is a Python library used for data analysis and manipulation.

It provides two main data structures:

Series – 1-dimensional labeled array.

DataFrame – 2-dimensional labeled table (like Excel).

**Pandas Series**

A Series is similar to a list or a single column in Excel but with labels (index).

Can store integers, floats, strings, or even mixed data.

Key Features
Has an index (row labels).

Can be created from a list, dictionary, or array.

Supports vectorized operations (fast calculations).

Example Use Cases:

Storing sales numbers for a week.

Keeping a list of product prices.



```python
import pandas as pd

# Creating a Series from a list
sales = [200, 300, 250, 400, 500]
# data=pd.Series(sales)
# print(data)

sales_series = pd.Series(sales, index=["Mon", "Tue", "Wed", "Thu", "Fri"])

print("Sales Series:")
print(sales_series)
print("\nAccessing a single value (Wed):", sales_series["Wed"])
```

**Pandas DataFrame**

A DataFrame is like an Excel spreadsheet or SQL table.

Consists of rows and columns, where each column is a Series.

Key Features
Can be created from dictionaries, lists of dictionaries, or CSV/Excel files.

Has both row index and column labels.

Supports filtering, grouping, merging, and more.

Example Use Cases:

Storing employee data (Name, Department, Salary).

Analyzing large datasets (sales, customer records).
