#**Print statement with \n and \t**

In Python, the \n and \t are escape sequences used within strings to add special formatting:

\n - Newline Character
The \n escape sequence creates a newline. It moves the cursor to the next line when printing.

\t - Tab Character
The \t escape sequence inserts a tab space, which is equivalent to pressing the Tab key.

```python
print("Hello \nworld \t Bye")  #Seperated by comman
```

#**f-strings in Python (formatted string literals)**

The f-strings in Python (introduced in Python 3.6) are a convenient way to embed expressions inside string literals, using curly braces {}. They allow for cleaner and more readable string formatting.

```python
name = "John"
age = 25
print("My name is",name, "and I am ",age,"years old")
print(f"My name is {name} and I am {age} years old.")
```

```python
#Embedding Expressions
a = 5
b = 3
print(f"The sum of {a} and {b} is {a + b}.")
```
