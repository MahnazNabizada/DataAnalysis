#**String Slicing**

```python
# # #String Slicing
# print(strdata)
# print(strdata[:]) #all the characters of a string
# print(strdata[0:5]) #All the characters from first till 4th -0,1,2,3,4
# print(strdata[0:7:2]) #Every alternate character between 0-7
print(strdata[-1:-5:-1]) #Last four characters
print(strdata[-4:])
print(strdata[5:])
print(strdata[-1:]) #Only last character beacuse negative jump is not provided
print(strdata[::]) #The entire string
print(strdata[2::2]) #Every second character from a string ( jump is 2 , last is not given )
print(strdata[-1:-4])  #It will not print anything because by default jump is considered +1
print(strdata[-1:-4:-1]) #It will print character -1,-2,-3
print(strdata[::-1]) #Reverse the string
print(strdata[-5:-1])# Print the characters -4,-3,-2,-1
print(strdata[:6]) # Retreive first six characters of the string
```
