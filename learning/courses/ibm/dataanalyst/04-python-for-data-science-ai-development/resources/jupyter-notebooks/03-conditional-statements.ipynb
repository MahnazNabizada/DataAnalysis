#**if-elif-else Statement**

Use this when you have multiple conditions to check.

```python
#Evaluate Grade based on Average score.
#If Total score> 80: Grade A ,
#Total score 60-80, Grade B ,
# 40-60, Grade C,
#less than 40: Grade F

score=int(input("Enter total score:"))
if score>=80:
  print("Grade A")
elif score>=60 and score<80:
  print("Grade B")
elif score>=40 and score<60:
  print("Grade C")
else:
  print("Grade F")

```

```python
#Other way of solving the same question

score = int(input("Enter total score: "))

if score >= 80:
    print("Grade A")
elif score >= 60:
    print("Grade B")
elif score >= 40:
    print("Grade C")
else:
    print("Grade F")
```

```python
#Using logical grouping to solv ethe same question

score = int(input("Enter total score: "))

if score >= 80:
    grade = "Grade A"
elif 60 <= score < 80:
    grade = "Grade B"
elif 40 <= score < 60:
    grade = "Grade C"
else:
    grade = "Grade F"

print(grade)
```

#**Nested if Statements Example**

In a nested if statement, you place one if statement inside another if statement to check for multiple conditions. Here's an example that checks if a person is eligible for a driver's license based on their age and whether they have passed a driving test.

Problem: Check if a person is eligible for a driver's license based on age and driving test result.


**Conditions:**

The person must be at least 18 years old.

If they are between 18 and 25 years old, they must  passed g2 driving test.
If they are older than 25, they can automatically eligible for  G  driver's license without passing the test.


```python
age = int(input("Enter your age: "))

if age >= 18 and age <= 25:  # Nested if for people between 18 and 25
        has_passed_test = input("Have you passed the driving test? (yes/no): ").lower()
        if has_passed_test == "yes":
            print("You are eligible for a driver's license!")
        else:
            print("You must pass the driving test to be eligible.")
elif age>25:
        print("You are eligible for a driver's license!")
else:
    print("You are not eligible for a driver's license yet.")
```
